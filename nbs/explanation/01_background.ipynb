# Where trufl comes from

> The research behind trufl and its place in the DSS4NAFA ecosystem

## DSS4NAFA

DSS4NAFA is a decision support system for nuclear emergencies affecting food and agriculture. The [Joint FAO/IAEA Centre of Nuclear Techniques in Food and Agriculture](https://www.fao.org/agriculture/fao-iaea-nuclear-techniques/en) led its development through two IAEA Coordinated Research Projects (CRPs):

- D15015, [Response to Nuclear Emergencies Affecting Food and Agriculture](https://www.iaea.org/fr/node/285860), from 2013 to 2019.
- D15019, [Monitoring and Predicting Radionuclide Uptake and Dynamics for Optimizing Remediation of Radioactive Contamination in Agriculture](https://www.iaea.org/newscenter/news/new-crp-monitoring-and-predicting-radionuclide-uptake-and-dynamics-for-optimizing-remediation-of-radioactive-contamination-in-agriculture-crp-d15019). trufl started within this CRP.

After a release of radionuclides, food and health authorities use DSS4NAFA to manage data on contaminated soil and crops. It also advises them on sampling and plans the sampling campaigns.

The [DSS4NAFA video](https://www.youtube.com/watch?v=Ut4GzjKabMc) shows the system. Lee Zhi Yi and Dercon (2019) describe the data management and visualisation approach behind it.

## How trufl fits in

trufl adds adaptive sampling strategies to the sampling plans of DSS4NAFA. At each round, trufl ranks the areas by sampling need from the measurements collected so far, then splits the sampling budget across them. [How trufl ranks areas](02_ranking.ipynb) explains the ranking.

DSS4NAFA imports a sampling plan as a CSV file with one row per sample location.

## Research context

trufl draws on the PhD research of Floris Abrams at KU Leuven and SCK CEN (Abrams, 2023). The thesis develops tools for deciding where, how and when to remediate agricultural land after a large-scale release of radionuclides. It applies them to two agricultural watersheds: the Maarkebeek in Flanders, Belgium, and the Niida in Fukushima Prefecture, Japan.

trufl also draws on OREFA, an unpublished toolbox from the same research.
