## Why trufl

Sampling theory and the research literature cover multi-stage and adaptive spatial sampling well. Their methods rest on assumptions, such as a planned probability sampling design or a known model of spatial variation. In a nuclear emergency, the data rarely meet those assumptions. Measurements arrive over time from many sources, such as airborne surveys, laboratory analyses and samples taken where it was convenient to take them. They cover different locations at different resolutions. Sound statistical inference from such data is difficult. Decisions still have to be made, and you can't ignore the data you have. trufl takes a pragmatic approach and lets you use it.

The reasons to sample one area before another also vary, and they change during an emergency. They include the levels measured so far, the uncertainty that remains, prior knowledge, land use and population. That is why trufl leaves the criteria and their weights to you, and lets you change both from one round to the next.

Like DSS4NAFA, trufl is not a black box. Each rank follows from the state of the area, the criteria, the weights and the scoring method. You can inspect each of them, and adjust them as the situation develops.

The same reasoning applies before an emergency, when a campaign can be rehearsed on a map of a possible contamination, and to other phenomena monitored over time, such as legacy contaminants.

## DSS4NAFA

DSS4NAFA is a decision support system for nuclear emergencies affecting food and agriculture. The [Joint FAO/IAEA Centre of Nuclear Techniques in Food and Agriculture](https://www.fao.org/agriculture/fao-iaea-nuclear-techniques/en) led its development through two IAEA Coordinated Research Projects (CRPs):

- D15015, [Response to Nuclear Emergencies Affecting Food and Agriculture](https://www.iaea.org/fr/node/285860), from 2013 to 2019.
- D15019, [Monitoring and Predicting Radionuclide Uptake and Dynamics for Optimizing Remediation of Radioactive Contamination in Agriculture](https://www.iaea.org/newscenter/news/new-crp-monitoring-and-predicting-radionuclide-uptake-and-dynamics-for-optimizing-remediation-of-radioactive-contamination-in-agriculture-crp-d15019). trufl started within this CRP.

After a release of radionuclides, food and health authorities use DSS4NAFA to manage data on contaminated soil and crops. It also advises them on sampling and plans the sampling campaigns.

The [DSS4NAFA video](https://www.youtube.com/watch?v=Ut4GzjKabMc) shows the system. Lee Zhi Yi and Dercon (2019) describe the data management and visualisation approach behind it.

## How trufl fits in

trufl adds adaptive sampling strategies to the sampling plans of DSS4NAFA. DSS4NAFA manages the campaign and its data. trufl addresses one question within it: given what is known now, where should the next samples go?

## Research context

trufl draws on the PhD research of Floris Abrams at KU Leuven and SCK CEN (Abrams, 2023). The thesis develops tools for deciding where, how and when to remediate agricultural land after a large-scale release of radionuclides. It applies them to two agricultural watersheds: the Maarkebeek in Flanders, Belgium, and the Niida in Fukushima Prefecture, Japan.

trufl also draws on OREFA, an unpublished toolbox from the same research.
