## Focus against coverage

Every allocation trades two fears. If samples are spread evenly, the worst places are poorly known when decisions about them are due. If they are concentrated, a mistake in the prior is never found out, because nobody goes where the prior said there was nothing.

trufl offers two degrees of concentration. Quartiles give groups that are easy to explain in a meeting, at the cost of cliffs: two neighbours in the list can receive very different numbers. Weighting by 1/rank has no cliffs but is steep. Over 100 areas, the first one alone takes about a fifth of the budget.

A minimum per area is insurance against a wrong prior. It also has a value that is not statistical: a community where nobody came to sample has little reason to trust the decision made about it.

## Why the total is not the budget

You might expect a budget of 20 to give 20 samples. Over 100 areas it can give none, or 100:

```python
#| hide
import warnings
warnings.filterwarnings('ignore')

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import geopandas as gpd
from fastcore.test import test_eq

from trufl.utils import gridder
from trufl.callbacks import State, MaxCB, MinCB, StdCB, CountCB, PriorCB
from trufl.optimizer import Optimizer
from trufl.sampler import Sampler, rank_to_sample
from trufl.collector import DataCollector
```

```python
#| echo: false
every_rank = np.arange(1, 101)
plans = pd.DataFrame({'samples planned': [rank_to_sample(every_rank, 20, m, 'quantiles').sum() for m in (0, 1)]},
                     index=pd.Index(['no minimum', 'one per area'], name='budget of 20'))
test_eq(plans['samples planned'].tolist(), [0, 100])
plans
```

Both come from one property: each area's number is worked out alone, and nothing afterwards adds them up.

- Without a minimum, every share is less than one sample, and each is rounded down to zero.
- With a minimum, every area is raised to one, and nothing is taken away elsewhere to pay for it.

Larger budgets hide the effect without removing it. This is why a total has to be checked each time, and why there is a [guide to fitting a plan to its budget](../how-to/04_fit_budget.ipynb).

## Why points are random

Within an area trufl has no reason to prefer one spot, and choosing at random protects against a planner's unconscious preference for the roadside or the familiar field. Chance has its own costs. Points cluster, leave holes and can miss a small hotspot entirely. Finding nothing in an area with one random sample is very weak evidence that nothing is there.

Randomness also assumes that all the ground inside an outline is equally available. It never is, which is why closed ground has to leave the outline before the points are drawn, and not the points afterwards.
