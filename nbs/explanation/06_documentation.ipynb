# How this documentation works

> Four kinds of page for four needs, explained with trufl as the example

::: {.callout-note appearance="simple" icon=false}
**Explanation: understand why.** Read it away from the work. There is nothing to run. Want to act? Use a [how-to guide](../how-to/02_use_your_data.ipynb). [About the four kinds of page](../explanation/06_documentation.ipynb).
:::

This documentation follows [Diátaxis](https://diataxis.fr), a way of organising documentation devised by Daniele Procida. Its starting point is that people come to documentation in different situations, and that a page written for one situation serves the others badly.

Imagine three people opening the trufl documentation on the same morning. A new colleague wants to learn. A planner has teams leaving at two o'clock. A reviewer asks why a village was skipped. One page for all three would make the colleague choose between options they don't understand, make the planner scroll through theory and bury the reviewer's answer in code.

## Two questions

Diátaxis sorts any piece of writing with two questions.

1. Is the reader **doing** something, or **thinking**?
2. Are they **learning**, or **working**?

| | Learning | Working |
|---|---|---|
| **Doing** | **Tutorial**: "teach me" | **How-to guide**: "help me get this done" |
| **Thinking** | **Explanation**: "help me understand" | **Reference**: "tell me the fact" |

In a kitchen these are a first cooking lesson, a recipe, the label on a packet and a book about why cooking works.

The difference is easiest to see when the subject stays the same. All four of these pages are about ranking.

| Kind | Page | How it speaks | What it leaves out |

Open the four side by side. The facts overlap, and the voice, order and omissions are completely different. That difference is Diátaxis.

## What each kind promises

**A tutorial promises success.** The teacher makes every choice, so the learner can't go wrong. Each step has a result to check. That is why Getting started uses one map, one policy and one set of weights, and never mentions an alternative.

**A how-to guide promises a result.** It starts from what you already have, and may branch, because real work does. That is why [Fit a plan to your budget](../how-to/04_fit_budget.ipynb) has a section for when the budget is too small, which would only confuse a learner.

**Reference promises the truth.** It is neutral, complete and dull, and shaped like the software it describes. That is why [Errors and silent failures](../reference/03_errors.ipynb) is a set of tables.

**Explanation promises understanding.** It may have opinions, compare and digress. That is why [What a plan can and cannot tell you](05_limits.ipynb) argues, and contains nothing to run.

## The mistakes that are easiest to make

The commonest is to confuse a tutorial with a how-to guide, since both are lists of steps. The difference is not easy against advanced. It is learning against working. A tutorial is for the benefit of the person, who will throw the result away. A guide is for the benefit of the result.

The next is kindness in the wrong place: a paragraph of theory in a tutorial, or advice in a reference table. Each addition seems helpful and makes the page slightly worse at its one job. The cure is nearly always a link.

## Test your understanding

Where would each of these belong? Decide with the two questions, then open the answer.

**1.** "`min` defaults to 0."

::: {.callout-tip collapse="true"}
## Answer
Reference. The reader is working and needs a fact.
:::

**2.** "The airborne survey arrived in another coordinate system, and I have to plan today."

::: {.callout-tip collapse="true"}
## Answer
A how-to guide. The reader is working and needs to act. It is step 2 of *Use your own areas and survey*.
:::

**3.** "Why has our worst village dropped down the list this year?"

::: {.callout-tip collapse="true"}
## Answer
Explanation. The reader is thinking, to understand. The answer is that it is now well known, and it needs no code.
:::

**4.** "I joined the team on Monday and have never planned a campaign."

::: {.callout-tip collapse="true"}
## Answer
The tutorial. The reader is learning and should act. Resist sending them to the explanation first: doing comes before understanding.
:::

**5.** A sentence proposed for the tutorial: "You could also use TOPSIS, which measures the distance to the worst area."

::: {.callout-tip collapse="true"}
## Answer
Not in the tutorial, as it offers a choice and a theory to someone learning by doing. The name and formula go to reference. The comparison goes to explanation.
:::

## If you change this documentation

Diátaxis advises against big plans. Take one page, or one paragraph, ask the two questions of it, and move or cut whatever answers differently from the rest of its page. Publish that, then take the next.
